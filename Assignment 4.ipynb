** Q1 Stop sending emails after 3 users (Using for loop and break)

users = ["Ali", "Sara", "John", "Zara", "Mike"]

Output:

Email sent to Ali

Email sent to Sara

Email sent to John

Limit reached. Stop sending emails.**

```python
users = ["Ali", "Sara", "John", "Zara", "Mike"]
for user in users:
    print("Email sent to" ,user )
    if user == "John":
        break;
```

**Q2. Stop monitoring temperature if critical temperature reached (Using for loop and break)

temps = [22, 25, 30, 45, 28] critical = 40

Ouput:

Temperature: 22

Temperature: 25

Temperature: 30

Critical temperature reached! Stop monitoring.**

```python
temps = [22, 25, 30, 45, 28]
critical = 40
for temp in temps:
    if temp >= critical:
        print("Critical temperature reached! Stop monitoring.")
        break;
    print("Temperature:", temp)
```

**Q3. Calculate total price of items in a shopping cart (Using For loop)

cart = [100, 250, 50, 300]

Output:

Total Price: 700**

```python
cart = [100, 250, 50, 300]
total = 0
for carts in cart:
    total = total + carts
print("Total Price:",total)
```

**Q4. Skip customers with unpaid bills (Using For loop and continue)

customers = [ {"name": "Ali", "paid": True},

{"name": "Sara", "paid": False},

{"name": "John", "paid": True} ]

Output:

Access granted to: Ali

Access granted to: John**

```python
customers = [
    {"name": "Ali", "paid": True},
    {"name": "Sara", "paid": False},
    {"name": "John", "paid": True}
]

for customer in customers:
    if customer["paid"] == False:
        continue
    print("Access granted to:", customer["name"])
```

**Q6. Extract domain names from email list (Using for loop)

emails = ["ali@gmail.com", "sara@yahoo.com", "john@hotmail.com"]

Output:

domains: ['gmail.com', 'yahoo.com', 'hotmail.com']**

```python
emails = ["ali@gmail.com", "sara@yahoo.com", "john@hotmail.com"]

domains = []

for email in emails:
    domain = email.split("@")[1]
    domains.append(domain)

print("domains:", domains)
```

**Q8. Create a multiplication table using a nested for loop.

numbers = [2, 3, 4] multipliers = [1, 2, 3]

Expected Output:

2 x 1 = 2 2 x 2 = 4 2 x 3 = 6 3 x 1 = 3 3 x 2 = 6 3 x 3 = 9 4 x 1 = 4 4 x 2 = 8 4 x 3 = 12

Use a nested for loop.**

```python
numbers = [2, 3, 4] 
multipliers = [1, 2, 3]
for number in numbers:
    for multiplier in multipliers:
        print(f'{number} * {multiplier} = {number * multiplier}')
        
```

**Q7. Create a list of numbers, generate a new list containing only the numbers that are divisible by 3 or 5. (List Comprehension)**

```python
numbers = [1, 3, 5, 9, 10, 12, 15, 20, 22, 30]
new_list = []
for num in numbers:
    if num %3 == 0 or num %5 == 0:
        new_list.append(num)
        print(new_list)
```

```python
numbers = [1, 3, 5, 9, 10, 12, 15, 20, 22, 30]
new_list = [num for num in numbers if num % 3 == 0 or num % 5 == 0]
print(new_list)
```

**Q10. Count down from 5 to 1 and then display "Time's up!".

Expected Output:

5 4 3 2 1 Time's up!

Use a while loop.**

```python
count = 5

while count > 0:
    print(count, "Time's up!")
    count -= 1
```

**Q9. Keep asking the user for a password until the correct password is entered.

correct_password = "python123"

Expected behaviour:

Enter password: abc Incorrect password. Try again.

Enter password: 12345 Incorrect password. Try again.

Enter password: python123 Access granted!

Use a while loop.**

```python
correct_password = "python123"

password = ""

while password != correct_password:
    password = input("Enter password: ")

    if password != correct_password:
        print("Incorrect password. Try again.")

print("Access granted!")
```
